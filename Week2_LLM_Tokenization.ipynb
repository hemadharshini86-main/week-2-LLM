```python
import pandas as pd
from transformers import AutoTokenizer
```

```python
df = pd.read_csv("../dataset/Clean_txt.csv")

print(df)
```

```python
texts = df["Clean_Text"].dropna().tolist()

print("Number of texts:", len(texts))
```

```python
text = texts[0]

print("Original Text:")
print(text)
```

```python
bert = AutoTokenizer.from_pretrained("bert-base-uncased")

print("BERT tokenizer loaded successfully!")
```

```python
bert_tokens = bert.tokenize(text)

print("BERT TOKENS:")
print(bert_tokens)
```

```python
bert_ids = bert.encode(text, add_special_tokens=True)

print("BERT TOKEN IDS:")
print(bert_ids)
```

```python
bert_decode = bert.decode(bert_ids)

print("BERT DECODED TEXT:")
print(bert_decode)
```

```python
gpt2 = AutoTokenizer.from_pretrained("gpt2")

print("GPT-2 tokenizer loaded successfully!")
```

```python
gpt2_tokens = gpt2.tokenize(text)

print("GPT-2 TOKENS:")
print(gpt2_tokens)
```

```python
gpt2_ids = gpt2.encode(
    text,
    add_special_tokens=False
)

print("GPT-2 TOKEN IDS:")
print(gpt2_ids)
```

```python
gpt2_decode = gpt2.decode(gpt2_ids)

print("GPT-2 DECODED TEXT:")
print(gpt2_decode)
```

```python
print("Original Text:")
print(text)

print("\nBERT TOKENS:")
print(bert_tokens)

print("\nGPT-2 TOKENS:")
print(gpt2_tokens)

print("\nBERT TOKEN COUNT:", len(bert_tokens))
print("GPT-2 TOKEN COUNT:", len(gpt2_tokens))
```

```python
df = pd.read_csv("../dataset/Clean_txt.csv")

texts = df["Clean_Text"].dropna().tolist()

print("Number of texts:", len(texts))
```

```python
gpt2 = AutoTokenizer.from_pretrained("gpt2")
```

```python
results = []

for text in texts:
    b_tokens = bert.tokenize(text)
    g_tokens = gpt2.tokenize(text)

    results.append({
        "Original Text": text,
        "BERT Tokens": " ".join(b_tokens),
        "GPT-2 Tokens": " ".join(g_tokens),
        "BERT Count": len(b_tokens),
        "GPT-2 Count": len(g_tokens)
    })

comparison = pd.DataFrame(results)

print(comparison)
```

```python
comparison.to_csv(
    "../dataset/comparison.csv",
    index=False
)

print("comparison.csv created successfully!")
```
